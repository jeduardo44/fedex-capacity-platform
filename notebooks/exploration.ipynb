# Exploration

Ad-hoc exploration of the DuckDB warehouse. Run the pipeline first (`make all`).

```python
import duckdb
import pandas as pd

con = duckdb.connect("../data/warehouse.duckdb", read_only=True)
con.execute("SELECT table_schema, table_name FROM information_schema.tables ORDER BY 1, 2").df()
```

```python
demand = con.execute("SELECT * FROM marts.demand_weekly LIMIT 10").df()
demand
```

```python
import plotly.express as px

df = con.execute("SELECT period_date, region, total_packages FROM marts.demand_weekly").df()
px.line(df, x="period_date", y="total_packages", color="region")
```

```python
con.close()
```
